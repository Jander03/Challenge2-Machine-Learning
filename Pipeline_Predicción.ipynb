# Pipeline de Predicción: Estimación de Defunciones por Neumonía
**Tarea de Machine Learning Seleccionada:** Regresión Lineal (Aprendizaje Supervisado).
**Justificación:** Se elige la regresión lineal porque el objetivo es predecir un valor numérico continuo (cantidad de defunciones). Los datos históricos reportados por el MINSA muestran una tendencia de reducción aproximadamente lineal, lo que hace de este modelo un excelente punto de partida para estimar el escenario de 2026.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

X_años = np.array([2023, 2024, 2025]).reshape(-1, 1)

y_defunciones = np.array([113, 82, 51])

modelo_prediccion = LinearRegression()
modelo_prediccion.fit(X_años, y_defunciones)

X_futuro = np.array([2026]).reshape(-1, 1)
prediccion_2026 = modelo_prediccion.predict(X_futuro)

X_total = np.concatenate((X_años, X_futuro))
y_tendencia = modelo_prediccion.predict(X_total)

plt.figure(figsize=(9, 5))
plt.scatter(X_años, y_defunciones, color='#e74c3c', s=120, label='Datos Históricos (MINSA)')
plt.scatter(X_futuro, prediccion_2026, color='#2980b9', marker='s', s=120, label=f'Predicción ML 2026: {int(prediccion_2026[0])} casos')
plt.plot(X_total, y_tendencia, color='#7f8c8d', linestyle='--', linewidth=2, label='Línea de Tendencia (Regresión)')

plt.title('Tendencia y Predicción de Defunciones por Neumonía (< 5 años)', fontsize=14, pad=15)
plt.xlabel('Año de Observación', fontsize=11)
plt.ylabel('Número de Defunciones', fontsize=11)
plt.xticks(X_total.flatten())
plt.ylim(0, 130)
plt.legend()
plt.grid(True, linestyle=':', alpha=0.7)

print(f"--- RESULTADO DE LA PREDICCIÓN ---")
print(f"Basado en la tendencia lineal de las intervenciones actuales,")
print(f"el modelo estima que las defunciones se reducirán a {int(prediccion_2026[0])} en el año 2026.\n")

plt.show()
```
