# Pipeline de Razonamiento y Búsqueda: Asignación Heurística de Vacunas
Este notebook implementa un agente de búsqueda que explora el espacio de estados para asignar el stock limitado de vacunas contra el neumococo a distintas regiones, priorizando aquellas con mayor mortalidad histórica (heurística).

```python
vacunas_disponibles = 2226142
regiones_estado = [
    {'nombre': 'Región Norte', 'episodios_ira': 350000, 'mortalidad_historica': 45, 'vacunas_asignadas': 0},
    {'nombre': 'Región Centro', 'episodios_ira': 274978, 'mortalidad_historica': 12, 'vacunas_asignadas': 0},
    {'nombre': 'Región Sur', 'episodios_ira': 200000, 'mortalidad_historica': 38, 'vacunas_asignadas': 0}
]

def busqueda_asignacion_optima(regiones, stock):
    print(f"--- INICIO DE EXPLORACIÓN Y BÚSQUEDA --- \nStock inicial disponible: {stock} vacunas\n")

    frontera_busqueda = sorted(regiones, key=lambda x: x['mortalidad_historica'], reverse=True)

    for nodo in frontera_busqueda:
        if stock > 0:
            asignacion = min(stock, nodo['episodios_ira'] * 2)
            nodo['vacunas_asignadas'] = asignacion
            stock -= asignacion
            print(f"-> Acción ejecutada: Asignando {asignacion} dosis a {nodo['nombre']} (Prioridad por mortalidad: {nodo['mortalidad_historica']})")

    print(f"\n--- RESULTADO DEL PIPELINE ---")
    print(f"Stock de contingencia restante: {stock} vacunas.")
    for res in frontera_busqueda:
        print(f" - {res['nombre']}: {res['vacunas_asignadas']} vacunas aseguradas.")

busqueda_asignacion_optima(regiones_estado, vacunas_disponibles)
```
